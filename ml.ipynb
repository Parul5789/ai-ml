```python
import pandas as pd
import numpy as np
```

```python
from sklearn.datasets import fetch_openml

boston = fetch_openml(name="boston", version=1, as_frame=True)
df = boston.frame
df.head()
```

```python
df.shape
```

```python
df.describe()
```

```python
import seaborn as sns
sns.displot(df['MEDV'])
```

```python
df.drop_duplicates()
```

```python
df.info()
```

```python
df.corr()['MEDV']
```

```python
x=pd.DataFrame(df[['LSTAT' , 'RM']])  #featue
y=pd.DataFrame(df['MEDV'])   #target
x
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test=train_test_split(x,y,test_size=0.2 , random_state=23)
```

```python
x_train
```

```python
y_train
```

```python
from sklearn.linear_model import LinearRegression
reg =LinearRegression();
reg.fit(x_train , y_train)
```

```python
print(reg.intercept_)
```

```python
reg.coef_
```

```python
y_pred=reg.predict(x_test)
y_pred=pd.DataFrame(y_pred , columns=['predictor'])
```

```python
y_pred
```

```python
res_df=pd.DataFrame(columns=['actual' , 'pred'])
res_df['actual']=y_test.reset_index()['MEDV']
res_df['pred']=y_pred.reset_index()['predictor']
res_df
```

```python
res_df['error']=res_df.actual-res_df.pred
```

```python
res_df
```

```python
from sklearn import metrics
print(metrics.mean_absolute_error(y_test , y_pred))
print(metrics.mean_squared_error(y_test , y_pred))
```

```python
metrics.mean_absolute_percentage_error(y_test , y_pred)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test , y_pred)
```
